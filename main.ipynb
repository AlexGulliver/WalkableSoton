```python
import osmnx as ox

# Optional: set logging and figure size
ox.settings.log_console = True
ox.settings.use_cache = True
ox.settings.default_figsize = (12, 12)

# Step 1: Download Southampton's bike-accessible road network
G = ox.graph.graph_from_place("Southampton, United Kingdom", network_type="bike")

# Step 2: Plot the graph
fig, ax = ox.plot_graph(G, node_size=0, edge_color="skyblue", edge_linewidth=1)
```
